```python
import os
import zipfile
import splitfolders
from ultralytics import YOLO
from google.colab import drive
zip_file_path = '/content/drive/MyDrive/Dataset_Hardware/dataset_garbage.zip'

unzip_folder = '/content/garbage_raw'

split_folder = '/content/garbage_ready'

print(f"Extracting file {zip_file_path}...")
os.makedirs(unzip_folder, exist_ok=True)

with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall(unzip_folder)

print("Extraction complete!")
target_folder_to_split = unzip_folder

for root, dirs, files in os.walk(unzip_folder):
    if len(dirs) >= 6:
        target_folder_to_split = root
        break
print("Dividing dataset into Train (80%) and Val (20%)...")
splitfolders.ratio(target_folder_to_split, output=split_folder, seed=1337, ratio=(.8, .2))
print(f"Data is ready at: {split_folder}")
```

```python
# =====================================================================
# 5. CONTINUE TRAINING YOLOV8 FROM CHECKPOINT (EPOCH 73)
# =====================================================================
print("Starting to continue training YOLOv8-cls...")

# 1. Point the direct path to the last.pt file of the interrupted training session
# It is usually located in the project/name/weights/ directory that you previously specified
last_weight_path = '/content/drive/MyDrive/Eco_Intelligence_Project/train_cls_run1-4/weights/last.pt'

# 2. Initialize the model from this last.pt file (DO NOT use 'yolov8n-cls.pt' anymore)
model = YOLO(last_weight_path)

# 3. Run the training command with the resume=True parameter
# YOLO will automatically reread the settings (epoch, imgsz, batch...) from the last.pt file
# and continue running from epoch 73 until reaching 500.
results = model.train(resume=True)

print("Training complete! Weight file (best.pt) has been saved to your Drive.")
```

```python
print("Starting training YOLOv8-cls...")
model = YOLO('yolov8n-cls.pt')

results = model.train(
    data=split_folder,
    epochs=100,
    imgsz=224,
    batch=32,
    device=0,
    project='/content/drive/MyDrive/Eco_Intelligence_Project',
    name='train_cls_224'
)

print("Training successfully!")
```
